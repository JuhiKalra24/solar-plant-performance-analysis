## Automating same across all months of plant 2

```python
monthly_sheets = [
    'Apr-25', 'May-25', 'Jun-2025', 'July-2025',
    'Aug-2025', 'Sep-2025', 'Oct-2025', 'Nov-2025'
]
```

```python
summary = []

for sheet in monthly_sheets:
    
    # extract right side table for analysis
    df_raw = pd.read_excel(file, sheet_name=sheet, header=None)
    
    # find right table start column
    date_cols = []
    for col in range(df_raw.shape[1]):
        if 'Date' in df_raw.iloc[:, col].astype(str).values:
            date_cols.append(col)
    
    right_col = date_cols[-1]
    df = df_raw.iloc[:, right_col:]
    
    # find header row
    for i in range(len(df)):
        if 'Date' in df.iloc[i].astype(str).values:
            header_row = i
            break
    
    df.columns = df.iloc[header_row]
    df = df[header_row + 1:]
    
    df.columns = df.columns.astype(str).str.strip()
    df = df.copy()
    
    # rename
    df = df.rename(columns={
        'Actual Generation (kWh)': 'Generation',
        'GHI kW/m2 Sat Data': 'GHI'
    })
    
    # convert numeric
    for col in df.columns:
        if col != 'Date':
            df[col] = pd.to_numeric(df[col], errors='coerce')
    
    df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
    df = df.dropna(subset=['Date'])
    
    # KPI
    total_gen = df['Generation'].sum()
    days = len(df)
    
    capacity = 1260  # confirm once
    cuf = total_gen / (capacity * 24 * days)

    # Approximate PR since GHI is available
    pr = total_gen / (df['GHI'].sum() * capacity)
    
    # active inverters
    inverter_cols = [col for col in df.columns if 'Inverter' in col]
    active_inv = df[inverter_cols].count().gt(0).sum()
    
    total_inv = len(inverter_cols)

    # Underperformance days
    underperf_days = df[df['Generation'] < 2000]

    num_underperf = len(underperf_days)

    # Classify using GHI
    underperf_days = underperf_days.copy()

    underperf_days['Reason'] = underperf_days['GHI'].apply(
      lambda x: 'Weather' if x < 3 else 'System'
    )

    weather_days = len(underperf_days[underperf_days['Reason'] == 'Weather'])
    system_days = len(underperf_days[underperf_days['Reason'] == 'System'])
    
    summary.append({
        'Month': sheet,
        'Generation': total_gen,
        'CUF': cuf,
        'PR': pr,
        'Active Inverters': active_inv,
        'Total Inverters': total_inv,
        'Underperformance Days': num_underperf,
        'System Issue Days': system_days,
        'Weather Issue Days': weather_days,
    })

    summary_df = pd.DataFrame(summary)
```
